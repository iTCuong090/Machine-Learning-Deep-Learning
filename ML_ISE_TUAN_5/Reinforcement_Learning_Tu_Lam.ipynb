# Reinforcement Learning

1. Trong Reinforcement Learning, hãy phân biệt state, action, reward, return và policy. Nêu vai trò của từng khái niệm trong quá trình agent học cách ra quyết định.?


State là trạng thái, thuộc tính hiện tại của agent và môi trường

Vai trò: Là cơ sở để Agent biết mình đang trong tình huống nào, phục vụ quyết định phải làm gì.


Action là các hành động mà agent có thể thực hiện tác động tới môi trường đó.

Vai trò: Việc thực hiện các action khác nhau có thể dẫn đến các state và reward khác nhau.


Reward là phần thưởng từ môi trường, đáp lại ngay lập tức cho agent sau hành động. Có thể dương, âm hoặc bằng 0.

Vai trò: Tín hiệu để báo hiệu cho Agent biết hành động vừa rồi là tốt hay xấu để điều chỉnh các hành động trong tương lai.

Return là tổng phần thưởng mà agent kì vọng sẽ nhận được kể từ một thời điểm trở đi.

Vai trò: Giúp agent đánh giá tác động lâu dài của hành động thay vì chỉ dựa trên reward tức thời.

Policy là một chiến lược đưa ra các Action mà agents sẽ tuân theo.

Vai trò: Dựa trên reward và return, điều chỉnh policy của agent để đem lại các kết quả đưa ra action tốt hơn.

2. State-action value function Q(s, a) biểu diễn điều gì? Giải thích trực giác của Bellman Equation và vì sao phương trình này là cơ sở để cập nhật ước lượng Q-value.

Q^pi(s,a) là giá trị kì vọng của tổng phần thưởng mà agent sẽ nhận được trong tương lai nếu đang ở trạng thái s, thực hiện hành động a và tiếp tục hành động theo policy pi.

Bellman Equation cho thấy giá trị này bằng reward hiện tại cộng giá trị chiết khấu của tương lai, nên nó được dùng làm mục tiêu để liên tục cập nhật và cải thiện ước lượng Q-value.

3. Discount factor γ ảnh hưởng như thế nào đến hành vi của agent? So sánh trường hợp γ gần 0 và γ gần 1, đồng thời giải thích sự đánh đổi giữa phần thưởng trước mắt và phần thưởng dài hạn.

Discount factor ảnh hưởng tới hành vi agent bằng cách điều chỉnh xu hướng chọn lợi ích ngay trước mắt hoặc lợi ích ở lâu dài. Nếu discount gần 0, agent có xu hướng lựa chọn những quyết định đem lại lợi ích ngay lập tức chú trọng reward, còn discount gần 1, agent có xu hướng lựa chọn những quyết định đem lại lợi ích dài hạn. Ví dụ trước mắt có vượt cửa A được 5 điểm nhưng có xu hướng bị trừ nhiều điểm nếu đi tiếp, nhưng vượt cửa B, C, D, E, F.. được 50 điểm thì agent sẽ bị lạc vào cửa A nếu discount thấp



4. Vì sao agent cần cân bằng exploration và exploitation? Giải thích cơ chế của ε-greedy policy và điều gì có thể xảy ra nếu ε luôn quá lớn hoặc quá nhỏ.

Exploration là việc agent khám phá những lựa chọn mà nó chưa bao giờ thực hiện, exploitation là việc agent vận dụng tối đa những gì nó đã biết để đạt được kết quả tốt nhất. Phải cân bằng giữa explore và exploit bởi vì có thể những khám phá mà agent chưa bao giờ thực hiện đem đến giải pháp tốt hơn những gì mà agent đang biết và đang nắm chắc, và cũng có thể nếu agent chỉ tập trung đi khám phá thì sẽ làm chậm quá trình training tích lũy kinh nghiệm và không đạt được kết quả mong muốn. E greedy policy là chiến lược tạo ra một xác suất mà agent sẽ đi explore hoặc đi exploit (chọn hành động có Q-value lớn nhất). Nếu e càng cao thì agent sẽ có xu hướng đi explore nhiều hơn, e càng nhỏ thì agent sẽ đi exploit nhiều hơn.


5. Khi state space là liên tục và quá lớn để lưu Q-table, Deep Q-Network (DQN) giải quyết vấn đề này như thế nào? Giải thích vai trò của neural network, experience replay/mini-batch và target network (soft update) trong việc giúp quá trình học ổn định hơn.

Khi state space là liên tục và quá lớn, ta sẽ tìm một hàm để biến những đầu vào liên tục từ state space sang đầu ra chiến lược hành động. Và hàm đó chính là một Neural network.
Không replay buffer: DQN vẫn có thể vừa chạy vừa train, nhưng mỗi kinh nghiệm thường chỉ được dùng ngay lúc xuất hiện → dữ liệu tương quan cao, dễ học không ổn định.
Có replay buffer: kinh nghiệm cũ được giữ lại và trộn ngẫu nhiên với kinh nghiệm mới → neural network học ổn định và tận dụng dữ liệu tốt hơn.
